**Project FinSafe: Data-Driven Coastal Safety**

Imagine traveling to Florida for the ultimate surfing trip or a dream beach vacation, stepping onto the sand, and having zero idea that you are standing in a documented shark activity hotspot. Every year, millions of tourists and water-sports enthusiasts enter the water completely unaware of local risks. We are here to change that before they paddle out.

**The Solution: Proactive, Data-Backed Awareness**

Project FinSafe is a Florida-based NGO dedicated to closing the information gap between ocean goers and marine safety. Instead of generic warnings, we combine marine wildlife tracking with precise data analytics to protect lives, support tourism, and promote safe coexistence with ocean wildlife.

**How We Do It**


- **Data-First Targeting:** We analyze historical attack documentation and regional records to pinpoint the exact beaches and coastal zones with the highest incident frequencies.
- **Targeted Interventions:** We bypass vague warnings by deploying striking, highly visible safety signs right at those high-risk access points.
- **On-the-Ground Education:** We partner with local surf schools, dive shops, and beachside hotels to distribute practical awareness flyers directly to tourists and practitioners who need it most.


**To drive our mission with precision, these are four critical questions to be answered using historical data:**

1. Which country should we operate in?
2. Which is the most dangerous location for sea activities?
3. Which is the least dangerous location for sea activities?
4. Which shark specimens are responsible for the most attacks?
5. What is the exact percentage breakdown of fatal, non-fatal, and unknown outcomes?


By answering these questions with hard data rather than guesswork, we ensure our signs, flyers, and educational campaigns are deployed precisely where they can save the most lives.

```python
#Database run
df = pd.read_excel(r'C:\Users\franc\Desktop\Ironhack\Week_2\Project\GSAF5.xls', engine='xlrd')
```

```python
#Duplicate drop - based on the full database
df = df.drop_duplicates()
```

```python
#General Data cleaning for fluid workflow
df = df[["Country","State", "Location", "Fatal Y/N"]]

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("/", "_")
)

df["country"] = df["country"].fillna("N/A")

df["state"] = df["state"].fillna("N/A")

df["location"] = df["location"].fillna("N/A")

df["country"] = df["country"].str.strip().str.replace("'", "").str.upper()

df["state"] = df["state"].str.strip().str.replace("'", "").str.upper()
```

```python
#Fatal Attacks Clean
unique_fatal = df["fatal_y_n"].unique()

df["fatal_y_n"] = df["fatal_y_n"].astype(str)
unique_fatal = df["fatal_y_n"].unique()

df["fatal_y_n"] = df["fatal_y_n"].str.strip().str.replace("'", "").str.upper()
unique_fatal = df["fatal_y_n"].unique()

fatal_mapping = {
    "Y": "Y",
    "N": "N",
    "M": "N",
    "NQ": "N",
    "F": "Unknown",
    "UNKNOWN": "Unknown",
    "2017": "Unknown",
    "Y X 2": "Unknown"
    
}

df["fatal_y_n"] = df["fatal_y_n"].replace(fatal_mapping)

df["fatal_y_n"] = df["fatal_y_n"].fillna("Unknown")
```

```python
#Value counts to understand where the company should operate & Convert to a DataFrame
top_countries_df = df['country'].value_counts().head(3).reset_index()
top_countries_df.columns = ['country', 'attack_count']

top_countries_df
```

```python
#Value counts for state
top_state_df = df['state'].value_counts().head(3).reset_index()
top_state_df.columns = ['state', 'attack_count']

top_state_df
```

```python
#Filtering the chosen country and location
filtered_df = df.loc[(df["country"] == "USA") & (df["state"] == "FLORIDA")]
filtered_df = filtered_df.reset_index(drop=True)
filtered_df
```

```python
#Showing the value count  for each location and the total sum
display(filtered_df["location"].value_counts())
print(f"The total amount of attacks in Florida State is: {filtered_df['location'].value_counts().sum()}")
```

```python
# Top 5 dangerous places
loc_counts = filtered_df["location"].value_counts()
total_attacks = loc_counts.sum()
top_5 = loc_counts.head(5)

top_5_df = top_5.reset_index()
top_5_df.columns = ['Location', 'Count']

top_5_df.insert(0, 'Rank', range(1, len(top_5_df) + 1))

display(top_5_df)
```

```python
#Top five dangerous locations - percentage wise
loc_counts = filtered_df["location"].value_counts()
total_attacks = loc_counts.sum()
top_5 = loc_counts.head(5)

print("The locations with the most shark attacks reported are:")

i = 1
for location, count in top_5.items():
    percentage = (count / total_attacks) * 100
    print(f"Top {i} = {location}: {percentage:.2f}%")
    i += 1
```

```python
#Random safest locations from the lowest valu
loc_counts = filtered_df["location"].value_counts()
target_counts = [5, 4, 3, 2, 1]
random_list = []

for count in target_counts:
    matching_locs = loc_counts[loc_counts == count]
    

    if len(matching_locs) > 0:
        random_loc = matching_locs.sample(n=1)
        random_list.append(pd.DataFrame({'Location': random_loc.index, 'Count': random_loc.values}))

if random_list:
    random_table = pd.concat(random_list, ignore_index=True)
    display(random_table)
else:
    print("No matching locations found for those counts.")
```

```python
# Top safest places - Percentage wise
random_dict = random_table.set_index('Location')['Count'].to_dict()

print("The locations with the lowest shark attacks reported are:")

i = 1
for location, count in reversed(random_dict.items()):
    percentage = (count / total_attacks) * 100
    print(f"Top {i} = {location}: {percentage:.2f}%")
    i += 1
```

```python
#Dangerous - Fatal and non-fatal percentage
top_5_dangerous = filtered_df["location"].value_counts().head(5).index
df_dangerous = filtered_df[filtered_df["location"].isin(top_5_dangerous)]

pivot_dangerous = df_dangerous.pivot_table(
    index="location",
    columns="fatal_y_n",
    aggfunc="size",
    fill_value=0
)

pivot_dangerous = pivot_dangerous.reindex(columns=['N', 'Y', 'Unknown'], fill_value=0)

pivot_dangerous_pct = (pivot_dangerous.div(pivot_dangerous.sum(axis=1), axis=0) * 100).round(2)

print("Top 5 Dangerous Locations - Outcome Percentages:")
display(pivot_dangerous_pct)
```

```python
#Safe - Fatal and non-fatal percentage
top_5_safest = filtered_df["location"].value_counts().sort_values(ascending=True).head(5).index
df_safest = filtered_df[filtered_df["location"].isin(top_5_safest)]

pivot_safest = df_safest.pivot_table(
    index="location",
    columns="fatal_y_n",
    aggfunc="size",
    fill_value=0
)

# Force the columns to always include N, Y, and UNKNOWN (filling missing ones with 0)
pivot_safest = pivot_safest.reindex(columns=['N', 'Y', 'Unknown'], fill_value=0)

pivot_safest_pct = (pivot_safest.div(pivot_safest.sum(axis=1), axis=0) * 100).round(2)

print("Top 5 Safest Locations - Outcome Percentages:")
display(pivot_safest_pct)
```
